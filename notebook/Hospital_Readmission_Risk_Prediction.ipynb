- Extract the three mapping sections

```python
blank_rows = map_df.index[
    map_df['admission_type_id'].isna() &
    map_df['description'].isna()
].tolist()


admission_map = map_df.loc[:blank_rows[0] - 1].dropna()

discharge_mapping = map_df.loc[
    blank_rows[0] + 2:blank_rows[1] - 1
].dropna()

source_mapping = map_df.loc[
    blank_rows[1] + 2:
].dropna()
```

- Create mapping dictionaries

```python
admission_dict = dict(zip(
    admission_map['admission_type_id'],
    admission_map['description']
))

discharge_dict = dict(zip(
    discharge_mapping['admission_type_id'],
    discharge_mapping['description']
))

source_dict = dict(zip(
    source_mapping['admission_type_id'],
    source_mapping['description']
))
```

- Add the mapped columns to your main dataset

```python
df['admission_type'] = df['admission_type_id'].map(admission_dict)

df['discharge_disposition'] = df['discharge_disposition_id'].map(discharge_dict)

df['admission_source'] = df['admission_source_id'].map(source_dict)
```

```python
from sklearn.model_selection import GroupShuffleSplit
```

```python
# Keep patient IDs for group splitting
groups = df['patient_nbr']

# Target variable
y = df['readmission_risk']

# Input features
X = df.drop(columns=[
    'encounter_id',
    'patient_nbr',
    'readmitted',
    'readmission_risk'
])

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Groups:", groups.nunique())
```

```python
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(
    splitter.split(X, y, groups=groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
train_patients = set(groups.iloc[train_idx])
test_patients = set(groups.iloc[test_idx])

print("Common patients:", len(train_patients.intersection(test_patients)))
```

```python
missing_percent = (
    X_train.isnull().mean() * 100
).sort_values(ascending=False)

print(missing_percent[missing_percent > 0].round(2))
```

```python
print(missing_percent[missing_percent > 0].round(2))
```

```python
numeric_features = X_train.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=['object']
).columns.tolist()

print("Numerical columns:", len(numeric_features))
print(numeric_features)

print("\nCategorical columns:", len(categorical_features))
print(categorical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Numerical columns
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine both
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
# Fit on training data and transform training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the same fitted pipeline
X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
# 1. Logistic Regression
log_model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

# 2. Decision Tree
dt_model = DecisionTreeClassifier(
    class_weight='balanced',
    random_state=42
)

# 3. Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

# Train all three models
log_model.fit(X_train_processed, y_train)
print("Logistic Regression trained.")

dt_model.fit(X_train_processed, y_train)
print("Decision Tree trained.")

rf_model.fit(X_train_processed, y_train)
print("Random Forest trained.")
```

```python
# Predictions from all three models
log_pred = log_model.predict(X_test_processed)

dt_pred = dt_model.predict(X_test_processed)

rf_pred = rf_model.predict(X_test_processed)

print("Predictions generated successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
models = {
    'Logistic Regression': (log_model, log_pred),
    'Decision Tree': (dt_model, dt_pred),
    'Random Forest': (rf_model, rf_pred)
}

results = []

for name, (model, predictions) in models.items():
    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, predictions),
        'Precision': precision_score(y_test, predictions, zero_division=0),
        'Recall': recall_score(y_test, predictions, zero_division=0),
        'F1 Score': f1_score(y_test, predictions, zero_division=0),
        'ROC-AUC': roc_auc_score(
            y_test, model.predict_proba(X_test_processed)[:, 1]
        )
    })

results_df = pd.DataFrame(results)

show_table(results_df.round(4))
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import joblib

# Load trained model and preprocessor
saved_data = joblib.load('readmission_model.pkl')

model = saved_data['model']
preprocessor = saved_data['preprocessor']

st.set_page_config(
    page_title="Hospital Readmission Risk Prediction",
    page_icon="🏥",
    layout="wide"
)

st.title("Hospital Patient Readmission Risk Prediction")

st.write(
    "Upload a CSV file containing patient encounter features "
    "to predict readmission within 30 days."
)

uploaded_file = st.file_uploader(
    "Upload Patient CSV",
    type=['csv']
)

if uploaded_file is not None:
    data = pd.read_csv(uploaded_file)

    # Remove identifiers and target columns if present
    data = data.drop(
        columns=[
            'encounter_id',
            'patient_nbr',
            'readmitted',
            'readmission_risk'
        ],
        errors='ignore'
    )

    try:
        processed_data = preprocessor.transform(data)
        predictions = model.predict(processed_data)

        results = data.copy()

        results['Predicted_Readmission'] = [
            'Within 30 Days' if value == 1
            else 'Not Within 30 Days'
            for value in predictions
        ]

        st.subheader("Prediction Results")
        st.dataframe(results)

        st.download_button(
            "Download Predictions",
            results.to_csv(index=False),
            "readmission_predictions.csv",
            "text/csv"
        )

    except Exception as e:
        st.error(f"Prediction error: {e}")
        st.info(
            "Please upload a CSV containing all required "
            "model input columns."
        )

st.warning(
    "Educational project only. Predictions are not intended "
    "for clinical decisions or patient care."
)
```
